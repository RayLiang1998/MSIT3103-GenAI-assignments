# Assignment 5 Demo — Qwen3-8B-Instruct with Ollama

## Demo Goal
This notebook demonstrates the practical capabilities of **Qwen3-8B-Instruct** in a real-world business support setting.

The selected demo use case is:

**AI Logistics Planning and Operations Assistant**

The model is tested on four practical tasks:
1. Summarization
2. Operational planning
3. Comparative reasoning
4. Structured output

## Model
- Model: `qwen3:8b`
- Runtime: Ollama local inference

## Setup Instructions
Before running this notebook, make sure the following commands work in your terminal:

```bash
ollama pull qwen3:8b
ollama run qwen3:8b


````python
import requests

MODEL_NAME = "qwen3:8b"
OLLAMA_URL = "http://localhost:11434/api/generate"

def run_ollama(prompt: str, model: str = MODEL_NAME) -> str:
    payload = {
        "model": model,
        "prompt": prompt,
        "stream": False,
        "think": False,
        "options": {
            "num_predict": 220,
            "temperature": 0.3
        }
    }

    response = requests.post(OLLAMA_URL, json=payload, timeout=180)
    response.raise_for_status()
    data = response.json()
    return data["response"].strip()
````

## Demo 1 — Summarization

This task demonstrates the model’s ability to summarize operational context into a concise and usable business description.


````python
prompt_1 = """
You are an operations assistant.

Please summarize the following warehouse operations issue in 3 concise sentences.

Context:
The warehouse has recently experienced frequent delays in outbound shipments.
Workers are spending too much time walking between storage zones.
High-demand SKUs are not positioned near packing stations.
Several picking routes overlap, causing congestion during peak hours.
Managers want to reduce travel time, improve picking efficiency, and shorten order fulfillment time.

Output requirements:
- Use 3 complete sentences
- Keep the summary concise and professional
"""

response_1 = run_ollama(prompt_1)

print("=== Prompt 1 ===")
print(prompt_1)
print("\n=== Response 1 ===")
print(response_1)
````

## Demo 2 — Operational Planning

This task demonstrates Qwen3-8B-Instruct’s ability to generate practical operational recommendations for a warehouse environment.


````python
prompt_2 = """
You are a warehouse logistics planning assistant.

Design a practical strategy to reduce delays in a warehouse environment.

Context:
- Order fulfillment is frequently delayed
- Workers walk long distances between zones
- Inventory layout is inefficient
- Picking congestion happens during busy periods
- Management wants actionable recommendations

Output requirements:
- Provide exactly 4 recommendations
- Each recommendation should be 1 sentence
- Focus on operational improvements
- Use a professional tone
"""

response_2 = run_ollama(prompt_2)

print("=== Prompt 2 ===")
print(prompt_2)
print("\n=== Response 2 ===")
print(response_2)
````

## Demo 3 — Comparative Reasoning

This task demonstrates Qwen3-8B-Instruct’s ability to compare alternative warehouse improvement plans and provide a recommendation with justification.


````python
prompt_3 = """
You are a logistics strategy assistant.

Compare the following two warehouse improvement plans and recommend the better option.

Plan A:
- Move fast-selling items closer to packing stations
- Keep current picking process unchanged
- No staffing changes

Plan B:
- Reorganize inventory zones by order frequency
- Introduce zone-based picking
- Rebalance staff across shifts
- Redesign routes to reduce overlap

Output requirements:
- First write a short comparison paragraph
- Then provide a final recommendation under the heading: Recommendation
- Keep the answer professional and concise
"""

response_3 = run_ollama(prompt_3)

print("=== Prompt 3 ===")
print(prompt_3)
print("\n=== Response 3 ===")
print(response_3)
````

## Demo 4 — Structured Output

This task demonstrates Qwen3-8B-Instruct’s ability to follow structured-output requirements in a business operations scenario.


````python
prompt_4 = """
You are a business AI assistant.

Read the warehouse scenario below and return the answer in valid JSON only.

Scenario:
A warehouse wants to reduce delays, improve picking speed, and decrease worker travel time.
Common issues include poor SKU placement, picking congestion, and unbalanced worker assignments.

Return JSON with exactly these keys:
{
  "main_problem": "",
  "recommended_actions": [],
  "expected_benefits": []
}
"""

response_4 = run_ollama(prompt_4)

print("=== Prompt 4 ===")
print(prompt_4)
print("\n=== Response 4 ===")
print(response_4)
````
